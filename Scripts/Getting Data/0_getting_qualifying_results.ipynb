```python
# Wikipedia requests that bots provide a descriptive User-Agent
headers = {
    'User-Agent': 'WM2026-Scraper/1.0 (your_email@example.com)'
}
```

```python
all_matches = []

for region, url in urls.items():
    print(f"Scraping {region} from Wikipedia...")
    
    response = requests.get(url, headers=headers)
    
    if response.status_code == 200:
        soup = BeautifulSoup(response.text, 'html.parser')
        
        # Find all Wikipedia "Football Box" templates
        # Wikipedia renders matches either in <table class="fevent"> or <div class="footballbox">
        match_boxes = soup.find_all(['table', 'div'], class_=['fevent', 'footballbox'])

        # For checking            
        print(f"Found {len(match_boxes)} matches for {region}.")
        
        # 3. Extract data from each match box
        for match in match_boxes:
            try:
                # The exact tag classes can vary slightly, but these are the standard Wikipedia classes
                home_team = match.find(class_='fhome').get_text(strip=True)
                away_team = match.find(class_='faway').get_text(strip=True)
                score    = match.find(class_='fscore').get_text(strip=True)
                
                # Some matches might not have a date explicitly in the box, or it's in a previous row
                date_element = match.find(class_='fdate')
                date = date_element.get_text(strip=True) if date_element else "Unknown Date"
                
                all_matches.append({
                    'Region': region,
                    'Date': date,
                    'Home': home_team,
                    'Score': score,
                    'Away': away_team
                })
            except AttributeError:
                continue
                
        time.sleep(3)
        
    else:
        print(f"Failed to retrieve {region}. Status Code: {response.status_code}")

```

```python
# Save to CSV
df = pd.DataFrame(all_matches)
df.to_csv('Data/wm2026_quali_games.csv', index=False)
print(df.head())
```
