I now want to get some details or stats for each individual player, to compute some form of "skill"-index for the whole team, which I can use later in model. Thanks to the popular FIFA-game, there are some websites which have this kind of informations, for instance https://sofifa.com/  or https://www.fifaindex.com/.  Luckily for us, there is also already a full data set availabe [Kaggle](https://www.kaggle.com/datasets/rovnez/fc-26-fifa-26-player-data/data), which contains over 100 attributes of over 10,000 different players

```python
df_player_stats = pd.read_csv("../Data/FC26_20250921.csv")
df_player_stats.shape
df_player_stats.head()
```

```python
# Select only relevant columns
df_player_stats = df_player_stats[["player_id","short_name","long_name","player_url","nationality_id","nationality_name","overall","defending","value_eur","age"]]
```

```python
# save as .csv
df_player_stats.to_csv("../Data/player_stats_2026.csv",index=False,sep=";",encoding='utf-8-sig')
```
