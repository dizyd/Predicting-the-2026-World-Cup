## Scrape Players from each Team


```python
import requests
import numpy as np
import pandas as pd
import io
from bs4 import BeautifulSoup
```

```python
# The official Wikipedia page for the 2026 World Cup squads
url = 'https://en.wikipedia.org/wiki/2026_FIFA_World_Cup_squads'
```

```python
headers = {
    'User-Agent': 'WM2026-Squad-Scraper/1.0 (your_email@example.com)'
}
```

```python
response = requests.get(url, headers=headers)

if response.status_code == 200:
    print("Successfully connected to Wikipedia. Scraping squads...\n")
    soup = BeautifulSoup(response.text, 'html.parser')
    
    # Countries are listed in <h3> tags right above their squad tables
    country_headers = soup.find_all('h3')
    
    all_players = []

    for header in country_headers:
        # Extract the country name 
        country = header.get_text(strip=True)
        
        # The squad table is the very next <table> tag after the country header
        table = header.find_next('table')
        
        if table:
            # Read the HTML table into a Pandas DataFrame
            try:
                html_stream = io.StringIO(str(table))
                df = pd.read_html(html_stream)[0]
                
                # Check if it is a squad table (some Wikipedia tables are formatting/info boxes)
                if 'Player' in df.columns:
                    # Add a new column so we know which country the player plays for
                    df['Country'] = country
                    
                    # Optional: Clean up Wikipedia reference brackets like "[1]" or "(captain)"
                    df['Player'] = df['Player'].str.replace(r'\[.*?\]', '', regex=True)
                    df['Player'] = df['Player'].str.replace(r'\(captain\)', '', regex=True, case=False).str.strip()
                    
                    all_players.append(df)
                    print(f"✓ Scraped {len(df)} players for {country}")
            except Exception as e:
                print(f"Skipped a table for {country}: {e}")

    # Combine all the individual country tables into one massive DataFrame
    if all_players:
        final_df = pd.concat(all_players, ignore_index=True)
        
        # Reorder columns to put 'Country' first for easier reading
        cols = ['Country'] + [col for col in final_df.columns if col != 'Country']
        final_df = final_df[cols]
        
        # Save the master list to CSV
        final_df.to_csv('..\Data\wm2026_all_players.csv', index=False)
        print(f"\nSuccess! Total of {len(final_df)} players saved to 'wm2026_all_players.csv'.")
        
    else:
        print("Could not find any player tables.")
else:
    print(f"Failed to retrieve page. Status Code: {response.status_code}")
```
